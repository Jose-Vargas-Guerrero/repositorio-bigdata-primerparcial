```python
Ejercicio 1: Clasificación de edad

Enunciado: Solicitar o definir una edad y determinar la categoría correspondiente según los rangos establecidos (niño, adolescente, adulto o adulto mayor).

Lógica: Se usan condicionales if, elif y else para evaluar la edad y clasificarla según el rango en el que se encuentre.

```

```python
edad = 20

if edad < 13:
    print("niño")
elif edad >= 13 and edad <= 17:
    print("adolescente")
elif edad >= 18 and edad <= 59:
    print("adulto")
else:
    print("adulto mayor")
```

```python
resultado: adulto
```

```python
Ejercicio 2: Conversor de temperatura

Enunciado: Convertir una temperatura en grados Celsius a Fahrenheit utilizando la fórmula establecida.

Lógica: Se aplica la fórmula matemática F = (C × 9/5) + 32 para obtener el valor en Fahrenheit y luego se muestra el resultado.
```

```python
celsius = 25

fahrenheit = (celsius * 9/5) + 32

print("Temperatura en Fahrenheit:", fahrenheit)
```

```python
resultado: Temperatura en Fahrenheit: 77.0
```

```python
Ejercicio 3: Calculadora básica

Enunciado: Definir o solicitar dos números y una operación matemática (suma, resta, multiplicación o división) y ejecutar la operación correspondiente con validación en caso de división entre cero.

Lógica: Se usan condicionales para identificar la operación seleccionada y ejecutar el cálculo correspondiente, controlando el caso de división entre cero.
```

```python
numero1 = 10
numero2 = 5
operacion = "division"

if operacion == "suma":
    print(numero1 + numero2)
elif operacion == "resta":
    print(numero1 - numero2)
elif operacion == "multiplicacion":
    print(numero1 * numero2)
elif operacion == "division":
    if numero2 != 0:
        print(numero1 / numero2)
    else:
        print("Error: no se puede dividir entre cero")
```

```python
resultado: 2.0
```

```python
Ejercicio 4: Número par o impar

Enunciado: Definir o solicitar un número entero y determinar si es par o impar.

Lógica: Se utiliza el operador módulo % para verificar si el número es divisible entre 2, clasificándolo como par o impar según el resultado.
```

```python
numero = 7

if numero % 2 == 0:
    print("El número es par")
else:
    print("El número es impar")
```

```python
resultado: El número es impar
```

```python
Ejercicio 5: Tabla de multiplicar

Enunciado: Definir un número entero y mostrar su tabla de multiplicar del 1 al 12.

Lógica: Se usa un ciclo while con un contador que va multiplicando el número y aumentando hasta llegar a 12.
```

```python
numero = 5
contador = 1

while contador < 12:
    print(numero * contador)
    contador += 1
```

```python
resultado: 25
30
35
40
45
50
55
```

```python
Ejercicio 6: Contador de vocales

Enunciado: Definir una frase y contar cuántas vocales contiene, incluyendo mayúsculas y minúsculas.

Lógica: Se recorre la frase con un for y se verifica si cada letra es una vocal, acumulando el total.
```

```python
frase = "HOLA como estas"
vocales = 'aeiouAEIOU'
contador = 0

for letra in frase:
    if letra in vocales:
        contador +=1

print("vocales encontradas:", contador)
```

```python
resultado: vocales encontradas: 6
```

```python
Ejercicio 7: Promedio de calificaciones

Enunciado: Definir 5 calificaciones, calcular el promedio e indicar si el estudiante aprobó o reprobó (>= 70).

Lógica: Se suman las notas, se calcula el promedio y se evalúa con una condición para determinar el resultado.
```

```python
caliificacion1 = 70
caliificacion2 = 80
caliificacion3 = 70
caliificacion4 = 77
caliificacion5 = 90

suma = (caliificacion1 + caliificacion2 + caliificacion3 + caliificacion4 + caliificacion5)

promedio = suma / 5

if promedio >= 70:
    print('aprobado')
else:
    print('reprobado')

print('su promedio es: ', promedio)
```

```python
resultado: aprobado
su promedio es:  77.4
```

```python
Ejercicio 8: Números positivos, negativos y ceros

Enunciado: Definir 10 números y contar cuántos son positivos, negativos y ceros.

Lógica: Se recorren los números con un ciclo y se clasifican usando condiciones if, elif y else.
```

```python
numero1 = 13
numero2 = -2
numero3 = 0
numero4 = 12
numero5 = -12
numero6 = 7
numero7 = 0
numero8 = -5
numero9 = 18
numero10 = -2

positivo = 0
negativo = 0
ceros = 0

numeros = [
    numero1, numero2, numero3, numero4, numero5,
    numero6, numero7, numero8, numero9, numero10
]

for numero in numeros:
    if numero > 0:
        positivo += 1
    elif numero < 0:
        negativo += 1
    else:
        ceros += 1

print('numero positivos: ', positivo)
print('numero negativos: ', negativo)
print('numero cero: ', ceros)
```

```python
numero positivos:  4
numero negativos:  4
numero cero:  2
```

```python
Ejercicio 9: Mayor y menor de una lista

Enunciado: Crear una lista de números enteros, encontrar el número mayor, el menor y calcular la diferencia entre ambos.

Lógica: Se usa una lista de números y las funciones max() y min() para obtener el mayor y menor valor. Luego se resta para obtener la diferencia entre ambos.
```

```python
numeros = [10, 5, 30, 2, 18, 25]

mayor = max(numeros)
menor = min(numeros)

diferencia = mayor - menor

print("Mayor:", mayor)
print("Menor:", menor)
print("Diferencia:", diferencia)
```

```python
resultado: Mayor: 30
Menor: 2
Diferencia: 28
```

```python
Ejercicio 11: Función para calcular descuento

Enunciado: Crear una función que reciba un precio original y un porcentaje de descuento, y retorne el precio final.

Lógica: La función recibe dos valores, realiza el cálculo del descuento y retorna el precio final después de aplicar la reducción.
```

```python
def calcular_descuento(precio, descuento):
    precio_final = precio - (precio * descuento / 100)
    return precio_final

precio = 1000
descuento = 20

resultado = calcular_descuento(precio, descuento)

print("Precio final:", resultado)
```

```python
resultado: Precio final: 800.0
```

```python
Ejercicio 12: Validación de contraseña

Enunciado: Validar una contraseña verificando longitud mínima, presencia de números, mayúsculas y minúsculas.

Lógica: Se evalúan condiciones sobre el texto de la contraseña para verificar si cumple con todos los requisitos establecidos.
```

```python
contrasena = "Abc12345"

tiene_numero = False
tiene_mayuscula = False
tiene_minuscula = False

if len(contrasena) >= 8:
    for caracter in contrasena:
        if caracter.isdigit():
            tiene_numero = True
        if caracter.isupper():
            tiene_mayuscula = True
        if caracter.islower():
            tiene_minuscula = True

if len(contrasena) >= 8 and tiene_numero and tiene_mayuscula and tiene_minuscula:
    print("Contraseña válida")
else:
    print("Contraseña inválida")
```

```python
resultado: Contraseña válida
```

```python
# Ejercicio 13: Sistema simple de inventario

Enunciado: Crear una lista de productos con código, nombre, precio y cantidad disponible. Mostrar todos los productos, buscar uno por código y calcular el valor total del inventario.

Lógica: Se usa una lista de diccionarios para guardar los productos. Luego se recorren con un ciclo for para mostrarlos, buscar un código y calcular el total.
```

```python
productos = [
    {"codigo": 101, "nombre": "Laptop", "precio": 15000, "cantidad": 3},
    {"codigo": 102, "nombre": "Mouse", "precio": 500, "cantidad": 10},
    {"codigo": 103, "nombre": "Teclado", "precio": 1200, "cantidad": 5}
]

print("Lista de productos:")

for producto in productos:
    print(producto)

codigo_buscar = 102

for producto in productos:
    if producto["codigo"] == codigo_buscar:
        print("\nProducto encontrado:")
        print(producto)

total_inventario = 0

for producto in productos:
    total_producto = producto["precio"] * producto["cantidad"]
    total_inventario += total_producto

print("\nValor total del inventario:", total_inventario)
```

```python
Lista de productos:
{'codigo': 101, 'nombre': 'Laptop', 'precio': 15000, 'cantidad': 3}
{'codigo': 102, 'nombre': 'Mouse', 'precio': 500, 'cantidad': 10}
{'codigo': 103, 'nombre': 'Teclado', 'precio': 1200, 'cantidad': 5}

Producto encontrado:
{'codigo': 102, 'nombre': 'Mouse', 'precio': 500, 'cantidad': 10}

Valor total del inventario: 56000
```

```python
Ejercicio 14: Registro de ventas
Enunciado
Crear una lista de ventas donde cada venta tenga cliente, producto, cantidad y precio unitario. Calcular el total por venta, el total general vendido y el cliente con la compra más alta.

Lógica utilizada
Se utiliza una lista de diccionarios para almacenar las ventas. Se recorren los datos con un ciclo for para calcular los totales y encontrar la compra más alta.
```

```python
ventas = [
    {"cliente": "Juan", "producto": "Laptop", "cantidad": 1, "precio": 15000},
    {"cliente": "Ana", "producto": "Mouse", "cantidad": 2, "precio": 500},
    {"cliente": "Luis", "producto": "Teclado", "cantidad": 3, "precio": 1200}
]

print("Registro de ventas:\n")

total_general = 0
mayor_compra = 0
cliente_mayor = ""

for venta in ventas:
    total_venta = venta["cantidad"] * venta["precio"]

    print("Cliente:", venta["cliente"])
    print("Producto:", venta["producto"])
    print("Total venta:", total_venta)
    print("-------------------")

    total_general += total_venta

    if total_venta > mayor_compra:
        mayor_compra = total_venta
        cliente_mayor = venta["cliente"]

print("\nTotal general vendido:", total_general)
print("Cliente con la compra más alta:", cliente_mayor)
```

```python
Ejercicio 15: Limpieza básica de texto
Enunciado
Definir una frase, eliminar espacios innecesarios, convertir el texto a minúsculas, reemplazar espacios dobles y contar la cantidad de palabras.

Lógica utilizada
Se aplican métodos de texto para limpiar y organizar la frase. Luego se divide el texto en palabras para contarlas.
```

```python
texto = "   Hola    Mundo Python   "

print("Texto original:")
print(texto)

texto = texto.strip()

texto = texto.lower()

texto = texto.replace("  ", " ")

palabras = texto.split()

cantidad_palabras = len(palabras)

print("\nTexto limpio:")
print(texto)

print("\nCantidad de palabras:", cantidad_palabras)
```

```python
Texto original:
   Hola    Mundo Python   

Texto limpio:
hola  mundo python

Cantidad de palabras: 3
```
